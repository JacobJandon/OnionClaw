````python
# Load SICRY
import sys
sys.path.insert(0, '/path/to/OnionClaw')

import sicry
import json
import os
from dotenv import load_dotenv

load_dotenv()  # Load .env for API keys

print(f"SICRY v{sicry.__version__}")
print(f"LLM Provider: {sicry.LLM_PROVIDER}")
````

## 1. Verify Tor Connection

Before searching, verify that Tor is running and SICRY can reach it.

````python
# Check Tor status
tor_status = sicry.check_tor()
print(f"Tor Status: {'✓ Connected' if tor_status else '✗ Not reachable'}")

if not tor_status:
    print("\n[ERROR] Tor is not running.")
    print("Start Tor with: sudo tor &")
else:
    print("\nTor is ready. Proceeding with dark web search...")
````

## 2. Check Dark Web Search Engine Health

OnionClaw v2.3.0 includes 12+ verified dark web search engines. Check which are online.

````python
# Check health of all search engines
engines = sicry.check_search_engines()

print(f"\nSearch Engine Status ({len(engines)} total):")
print("\n" + "="*70)

online = [e for e in engines if e['status'] == 'up']
for e in online:
    latency = e.get('latency_ms')
    print(f"✓ {e['name']:25s} | {latency:4d}ms | Reliability: {e.get('reliability', 0):.1%}")

print(f"\n✓ {len(online)}/{len(engines)} engines online")

offline = [e for e in engines if e['status'] != 'up']
if offline:
    print(f"\n✗ {len(offline)} engines offline:")
    for e in offline:
        print(f"  - {e['name']}: {e.get('error', 'unknown')}")
````

## 3. Basic Dark Web Search

Search for information on the Tor network using mode-based routing.

````python
# Example 1: Threat Intelligence Search
query = "ransomware group infrastructure"

print(f"Searching: {query}")
print(f"Mode: threat_intel (auto-selects best engines)\n")

results = sicry.search(
    query=query,
    max_results=15,
    mode="threat_intel",  # Auto-selects engines for threat intel context
)

print(f"Found {len(results)} results\n")
print("="*70)

for i, r in enumerate(results[:5], 1):
    print(f"\n{i}. {r['title'][:70]}")
    print(f"   URL: {r['url'][:60]}...")
    print(f"   Engine: {r['engine']:15s} | Confidence: {r.get('confidence', 0):.2f}")
````

## 4. Deduplication (v2.3.0)

OnionClaw v2.3.0 automatically deduplicates results by:
- Removing exact URL matches
- Detecting domain mirrors (same content, different domain)

This happens internally; here's how to manually dedupe:

````python
# Manual deduplication example
print(f"Total results: {len(results)}")

# Intern function _deduplicate_results is called automatically in search()
deduped = sicry._deduplicate_results(results)

print(f"After dedup: {len(deduped)}")
print(f"Removed: {len(results) - len(deduped)} duplicates/mirrors\n")

# Show uniqueness
domains = set()
for r in deduped:
    import re
    m = re.match(r"https?://([^/]+)", r['url'])
    if m:
        domains.add(m.group(1))

print(f"Unique domains: {len(domains)}")
print("Dedup is automatic in search(); view coverage above.")
````

## 5. Semantic Re-Ranking with LLM (v2.3.0)

Optional: Use an LLM to re-score results by semantic relevance.

This improves precision but adds latency (LLM calls).

````python
# Enable semantic re-ranking (uses configured LLM provider)
query_to_rank = "critical vulnerability exploitation techniques"

# First: get raw results
raw_results = sicry.search(query_to_rank, max_results=10, _use_cache=False)

print(f"Raw Results (BM25 scoring):")
print(f"\n{'Rank':>4} {'Title':40s} {'Confidence':>12}")
print("-" * 60)
for i, r in enumerate(raw_results[:5], 1):
    title = r['title'][:37] + "..."
    conf = f"{r.get('confidence', 0):.2f}"
    print(f"{i:>4d} {title:40s} {conf:>12s}")

# Second: re-rank semantically
print(f"\n\nRe-ranking with LLM ({sicry.LLM_PROVIDER.upper()})...")
ranked = sicry._rank_results_semantic(query_to_rank, raw_results)

print(f"\nSemanticlly Ranked Results:")
print(f"\n{'Rank':>4} {'Title':40s} {'LLM Score':>12}")
print("-" * 60)
for i, r in enumerate(ranked[:5], 1):
    title = r['title'][:37] + "..."
    score = f"{r.get('confidence', 0):.2f}"
    print(f"{i:>4d} {title:40s} {score:>12s}")

print("\n✓ LLM semantic ranking improved result precision.")
````

## 7. Multi-Step OSINT Investigation

Real-world example: Investigate a reported compromise

````python
# Step 1: Threat intel search
investigation_query = "data exfiltration underground forums"
print(f"Investigation: {investigation_query}\n")

# Search forums/marketplaces
forums = sicry.search(
    investigation_query,
    max_results=20,
    mode="corporate",  # Optimized for corporate threat intel
)

print(f"Step 1: Found {len(forums)} locations discussing data exfiltration\n")

# Analyze top results
if forums:
    top_3 = forums[:3]
    print("Top Threat Vectors:")
    for i, forum in enumerate(top_3, 1):
        print(f"{i}. {forum['title'][:60]}")
        print(f"   Source: {forum['engine']:15s} | Confidence: {forum.get('confidence', 0):.2f}\n")
````

````python
# Step 2: Use LLM to analyze for actionable intelligence
if forums:
    analysis_prompt = f"""
You are a threat intelligence analyst. Based on these dark web forum results about {investigation_query}:

Top results:
""" + "\n".join(
        f"- {f['title'][:70]} (source: {f['engine']})"
        for f in forums[:5]
    ) + """

Provide:
1. Key threat actors or groups mentioned
2. Attack patterns identified
3. Mitigation recommendations
4. Suggested next-step hunts

Be concise and actionable for a SOC team.
    """
    
    analysis = sicry._call_llm(
        sicry.LLM_PROVIDER,
        "You are a cybersecurity threat analyst.",
        analysis_prompt
    )
    
    print("=" * 70)
    print("THREAT ANALYSIS")
    print("=" * 70)
    print(analysis)
    print("\n" + "=" * 70)
````

## 8. Scraping + LLM Analysis

Fetch actual page content from results and analyze with LLM (advanced).

````python
# fetch() retrieves content from .onion URLs via Tor
if forums:
    target_url = forums[0]['url']
    print(f"Fetching content from: {target_url}")
    
    content = sicry.fetch(target_url, timeout=30)
    
    if content and len(content) > 100:
        print(f"\n✓ Fetched {len(content)} bytes of content")
        print(f"\nPreview (first 300 chars):\n")
        print(content[:300])
        print("...")
    else:
        print("\n✗ Could not fetch content (page may be down or blocked)")
````

## 9. Integration Example: LangChain

Use SICRY as a LangChain tool in a multi-step reasoning agent.

````python
# SICRY can be wrapped as a LangChain Tool
from langchain.tools import Tool

sicry_tool = Tool(
    name="search_dark_web",
    func=lambda q: json.dumps(
        [{
            "title": r["title"],
            "url": r["url"],
            "engine": r["engine"],
            "confidence": round(r["confidence"], 2)
        } for r in sicry.search(q, max_results=10)]
    ),
    description="Search the Tor network for dark web information. Input: query string.",
)

print("✓ SICRY tool created for LangChain")
print(f"  Tool Name: {sicry_tool.name}")
print(f"  Description: {sicry_tool.description[:60]}...")

# Test tool
result = sicry_tool.run("ransomware payment tracking")
print(f"\nTool test successful! Returns JSON with {len(json.loads(result))} results.")
````

## 12. Pro Tips

1. **Use modes**: `search(query, mode="ransomware")` auto-selects best engines
2. **Check engine health first**: `check_search_engines()` before investigations
3. **Cache is your friend**: Results cached for 30 min; set `_use_cache=False` to force refresh
4. **Batch multiple searches**: Use different providers for latency comparison
5. **Combine data sources**: Use `fetch()` to get raw page content, then `ask()` for analysis
6. **Monitor reliability**: `engine_reliability_scores()` to track engine uptime
7. **Semantic ranking**: Only needed for small result sets; use for final 10-20 results
8. **Respect Tor**: Space out requests; rotate user agents; use reasonable timeouts
